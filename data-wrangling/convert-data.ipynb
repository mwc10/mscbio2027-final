## Download Data From Zenodo

They have provided [gigabytes of data at their zenodo repository](https://zenodo.org/records/14204217). Most of the data is in `RDS` format–aka raw R objects. For this project, it shouldn't be necessary to download all of the compressed archives. So far, I've used:

- [`Clinical.tar`](https://zenodo.org/records/14204217/files/Clinical.tar?download=1)
  - This has the sample information
- [`Robjects.tar`](https://zenodo.org/records/14204217/files/Robjects.tar?download=1)
  - Per-spot expert annotations (`annotsBySpot/TNBC<id>.RDS`)
  - Per-spot batch corrected gene counts (`counts/TNBC<id>.RDS`)
- [`classification.tar`](https://zenodo.org/records/14204217/files/classification.tar?download=1)
  - Results of the morphological category estimation for each spot

If you extract those two tar files into `data/ST/`, it will line up with the bash script below. I'd bet it's faster to download the files directly rather than through Jupyter curl.

If you download the files, extract the archives into the `data/ST/` directory if you want to simply run the conversion script.


```python
%%bash

base_dir='../data/ST'
output_dirs=('Clinical' 'Robjects' 'classification')
external_data=('https://zenodo.org/records/14204217/files/Clinical.tar?download=1' 'https://zenodo.org/records/14204217/files/Robjects.tar?download=1' 'https://zenodo.org/records/14204217/files/classification.tar?download=1')

# Make sure the arrays are the same length
if [ ${#output_dirs[@]} -ne ${#external_data[@]} ]; then
  echo "Error: output_dirs and external_data must have the same number of elements."
  exit 1
fi

# Ensure base directory exists
mkdir -p "$base_dir"


# Loop over pairs
for i in "${!output_dirs[@]}"; do
  TARGET_SUBDIR="${output_dirs[$i]}"
  TARGET_DIR="$base_dir/$TARGET_SUBDIR"
  TAR_URL="${external_data[$i]}"
  TAR_FILE="$(basename "${TAR_URL%%\?*}")"
  
  echo my file: $TAR_FILE

  # Step 1: Check if directory exists
  if [ ! -d "$TARGET_DIR" ]; then
    echo "Directory '$TARGET_DIR' not found. Downloading tar file from $TAR_URL..."

    # Step 2: Download tar file
    curl -L -o "$TAR_FILE" "$TAR_URL"

    # Step 3: Create target directory
    mkdir -p "$TARGET_DIR"

    # Step 4: Extract tar file into the directory
    tar -xzf "$TAR_FILE" -C "$base_dir"

    echo "Extraction to '$TARGET_DIR' complete."

    # Optional: Remove the tar file after extraction
    rm "$TAR_FILE"
  else
    echo "Directory '$TARGET_DIR' already exists. Skipping download and extraction."
  fi

  echo "------------------------------------"
done
```
